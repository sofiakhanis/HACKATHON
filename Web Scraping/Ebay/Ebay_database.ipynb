```python
import pandas as pd
import random
import uuid
import numpy as np
```

```python
brands = ['Samsung', 'LG', 'Sony', 'TCL', 'Hisense', 'Vizio', 'Panasonic', 'Sharp']
screen_sizes = [32, 40, 43, 50, 55, 58, 60, 65, 70, 75, 85]
smart_feature = ['✅', '❌']

def generate_product_name(brand, size):
    return f"{brand} {size}\" 4K Ultra HD TV"

def generate_link(product_id):
    return f"https://www.ebay.com/itm/{product_id}"
```

```python

data = []
for _ in range(400):
    brand = random.choice(brands)
    size = random.choice(screen_sizes)
    product_id = str(uuid.uuid4())[:8]
    name = generate_product_name(brand, size)
    price = round(random.uniform(129.99, 1899.99), 2)
    rating = round(random.uniform(2.5, 5.0), 1)
    rating_count = random.randint(10, 5000)
    link = generate_link(product_id)
    smart = random.choice(smart_feature)

    data.append({
        "product_id": product_id,
        "product_name": name,
        "product_brand": brand,
        "category": "TV",
        "price": price,
        "rating": rating,
        "rating_count": rating_count,
        "product_link": link,
        "smart_feature": smart,
        "screen_size_inch": size
    })


df_ebay = pd.DataFrame(data)
```

```python
df_ebay['smart_feature'] = df_ebay['smart_feature'].replace({'✅': 'Yes', '❌': 'No'})


df_ebay.head(10)
```

```python
df_ebay.to_csv("ebay_tv_synthetic.csv", index=False)


from google.colab import files
files.download("ebay_tv_synthetic.csv")
```
