```python
import pandas as pd
import random
import numpy as np
```

```python
#
brands = ["Samsung", "LG", "TCL", "Sony", "Hisense", "Vizio", "Philips", "Panasonic", "Sanyo", "Sharp"]
name_patterns = [
    "{brand} {size} Inch 4K Smart TV",
    "{brand} {size} Inch OLED TV",
    "{brand} {size} Inch Roku Smart TV",
    "{brand} {size} Inch Android TV",
    "{brand} {size} Inch LED Smart TV",
    "{brand} {size} Inch UHD TV",
    "{brand} {size} Inch HD Ready TV"
]
screen_sizes = list(range(32, 86))

tv_list = []
for i in range(1000):
    brand = random.choice(brands)
    size = random.choice(screen_sizes)
    name = random.choice(name_patterns).format(brand=brand, size=size)
    entry = {
        "product_id": f"TV{i+1000}",
        "product_name": name,
        "product_brand": brand,
        "category": "Electronics > TV & Video > Smart TVs",
        "price": round(random.uniform(150, 2500), 2),
        "rating": round(random.uniform(3.5, 5.0), 2),
        "rating_count": random.randint(10, 10000),
        "product_link": f"https://www.walmart.com/ip/{1000000 + i}",
        "smart_feature": random.choice(["Yes", "No"]),
        "screen_size_inch": size
    }
    tv_list.append(entry)


sample_tv_data = pd.DataFrame(tv_list)

```
