```python
import pandas as pd
import random

def generate_tv_data(n=300):
    brands = ["Samsung", "LG", "Sony", "TCL", "Insignia", "Vizio", "Hisense", "Panasonic", "Philips", "Sharp"]
    sizes = [24, 28, 32, 40, 43, 48, 50, 55, 58, 60, 65, 70, 75, 85]
    ratings = [round(r, 1) for r in list(set([random.uniform(4.1, 4.9) for _ in range(20)]))]

    data = []
    for i in range(n):
        brand = random.choice(brands)
        size = random.choice(sizes)
        rating = random.choice(ratings)
        rating_count = random.randint(50, 10000)
        price = round(size * random.uniform(5, 22), 2)
        smart = '✅' if random.random() > 0.5 else '❌'
        product_id = f"TVPROD{i:04d}"
        resolution = random.choice(["HD", "Full HD", "4K UHD", "8K UHD"])
        os = random.choice(["Android", "Roku", "Tizen", "WebOS", "Fire TV"])
        product_name = f"{brand} {size}\" {resolution} Smart LED TV with {os}"
        product_link = f"https://www.amazon.com/dp/{product_id}"

        data.append({
            "product_id": product_id,
            "product_name": product_name,
            "product_brand": brand,
            "category": "TV",
            "price": price,
            "rating": rating,
            "rating_count": rating_count,
            "product_link": product_link,
            "smart_feature": smart,
            "screen_size_inch": str(size)
        })

    return pd.DataFrame(data)

df_tv = generate_tv_data(300)

df_tv['smart_feature'] = df_tv['smart_feature'].map({'✅': 'Yes', '❌': 'No'})

from IPython.display import display
display(df_tv.head(10))

df_tv.to_csv("amazon_tv_data.csv", index=False, encoding='utf-8-sig')

from google.colab import files
files.download("amazon_tv_data.csv")
```
